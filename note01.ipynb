## Dimensionality reduction technique and their application in genomics and ecology

Many data analyses methods can be understood in a view that data to be
explained in a way to response to straight lines, representing real
variables or hypothetical latent variables. And the latent variables may
eventually be explained by some facts such as populations the samples
belong to, geographic locations. Regression is a way to explore the
relations between variables and their response variables, and the
least-squares is a way to do linear regression.

**1. Rectilinear response between $y_{ki}$ and explanatory variable**

To find the hypothetical explanatory variables, e.g. environmental variable, which species aboundance(or geno type values) linearly responds to. The heuristic utilize the least-squares approach can obtain eigenvalues/eigenvectors of **Y** iteratively. 

1\) To find a line Ey=b<sub>1\*</sub>x+e by estimating
$b_{1}$=$\sum_{i = 1}^{N}{(y_{i} - \overline{y})}$($x_{i}$-$\overline{x}$)/$\sum_{i = 1}^{N}{(x_{i} - \overline{x})}^{2}$ (suppose the relation between x and y can be described as a line y=b1\*x+e) and minimizing residual sum squares $\ \sum_{i = 1}^{N}{(y_{i} - Ey_{i})}^{2}$ (i.e. minimizes the sum of squared vertical distances). 
Usually, and in this notebook, we discuss centered data, that is to say
that both $\overline{x}$,$\overline{y}$ equal zero, unless we declare specifically. Then the
equation become $b_{1}$=$\sum_{i = 1}^{N}{y_{i}x_{i}/}\sum_{i = 1}^{N}{x_{i}}^{2}$.
$b_{k1}(or\ b_{k})$ for each specie k, can be estimated by such linear regression. 

![统计学习笔记_202505261707_08115 6.jpg](res/202505261816_259096.jpg)

The regression process is clearly illustrated above and easy to understand. When each y<sub>ki</sub> can't prefectly lie on the response line, the $A\mathbf{t}$=$\mathbf{y}$<sub>k</sub> is unsolvable and $\mathbf{y}$<sub>k</sub> is not in the C(A), the colume space of A. Finding the regression line is actually finding the projection of $\mathbf{y}$<sub>k</sub> on the C(A),Estimated $\mathbf{y}$<sub>k</sub>. i.e. to solve

$A^{T}A\mathbf{t}  =A^{T}\mathbf{y} $

2\) But different species may fit different explanatory variables better. To get an overall impression that achieve best fit to different explanatory variables for different species, the total regression sum of squares can be used to judge the goodness-of-fit. Then we got b<sub>k1</sub> for each specie. One can further do the regression reversely, i.e. estimate x through y by using the estimated b1, called ‘calibration’:

x<sub>i</sub>=$\sum_{k = 1}^{M}{y_{ki}b_{k}/\sum_{k = 1}^{M}b_{k}^{2}}$

3\) The difference between the estimated x<sub>i</sub> and the original/last x<sub>i</sub> could be used to measure the convergence of iterations. In gradient analysis, ordination do such multiple regression and calibration simultaneously. This process can be done through a two-way weighted algorithm iteration process. For the second and after latent variable, the trial scores are made uncorrelated with the first latent variable.


The same result can be obtained via matrix algebra through PCA, which constructs purely hypothetical axes that maximize the dispersion (variation) of sites’ projection on it (projecting each M-dims site point onto it), instead of fitting data to explanatory variables, such as environment variables, through regressions.

The projection of each site on PC1 (vectors consist of **P**) can also
be obtained through the iteration way: first, linear regression of each
species obtains slop b<sub>k1</sub>. The bigger b<sub>k1</sub>, the more
severe the abundance response to this species. Thus the ordination score
for each site was then obtained by weighted averaging all species’
b<sub>k1</sub> at that site. The b<sub>k1</sub> is also called the
species loading or species score for species k with respect to this
axis. The eigenvalue of this axis is equal to the goodness-of-fit, i.e.
the total sum of squares of regressions of the species abundances on the
axes. Minimized residual sum squares for overall can be done through the
iterative way.
