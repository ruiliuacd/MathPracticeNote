Whereas sometimes data is not a response to explanatory variables in a
linear relation. In gradient studies of ecology, there have been a lot
of heuristic methods to explore such situations. The key idea to achieve
such ordination that fits unimodal response relation is about the
arrangement of peak values’ coordination of all unimodal curves: first,
one can use the weighted averaging way to get the peak value for each
species, i.e. estimated species optimum/mode, usually denoted by
u<sub>k</sub> . Then one can calculate the ordinate scores x<sub>i</sub>
for each sample/site by weighted those u<sub>k</sub> . The goal is to
find the axes that maximum the dispersion of the species scores which
can be written as δ=$\sum_{k = 1}^{M}{y_{k +}u^{2}/y_{+ +}}$. The
complete iteration process is described below, called reciprocal
averaging that is an extension of two-way weighted methods.

Reciprocal averaging,

1.  Iteration process

> Step 1. Take arbitrary, but unequal, initial site scores x<sub>i</sub>
>
> Step 2. Calculate new species scores (u<sub>k</sub>) by weighted
> averaging of the site scores
>
> u<sub>k</sub>=$\sum_{i = 1}^{N}{y_{ki}x_{i}/\sum_{i = 1}^{N}y_{ki}}$
>
> Step 3. Through weighted averaging of u<sub>k</sub> to get new site
> scores
> x<sub>i</sub>=$\sum_{i = 1}^{M}{y_{ki}u_{k}/\sum_{i = 1}^{M}y_{ki}}$

1.  Orthogonalization procedure (not for the first axis)

2.  Standardization procedure

> Step 1. Calculate the centroid, z, of site scores {x<sub>i</sub>}.
> z=$\sum_{i = 1}^{N}{y_{+ i}x_{i}}/y_{+ +}$
>
> Step 2. Calculate the dispersion of site scores
> $s^{2} = \sum_{i = 2}^{N}{y_{+ i}{(x_{i} - z)}^{2}/y_{+ +}}$
>
> Step 3. Calculate $x_{i,new}$=${(x}_{i,old}$-z)/s

Then the expected response value, e.g. species abundance, under
bell-shape response model (parabola can also be log-transformed into
normal distribution), can be calculated by:

E($y_{ki}$)=${Ce}^{\frac{{{(x}_{i} - u_{k})}^{2}}{t_{k}^{2}}}$

Such two-way weighted approach is an iteration way to achieve
Correspondence Analysis (CA). A vector {x<sub>i</sub>} can be obtained
when the iteration finally convergent, then this vector is the axes to
project the data. That’s eigenvector (corresponding to the biggest
eigenvalue) and can essentially represent how it had been defined since
the scores remain the same in an extra iteration cycle. And all those
can be finally transformed into terms of algebra: (In CA, the matrix
**Y** is not previously centered.)

**u**={$u_{k}$}=$\mathbf{M}^{\mathbf{- 1}}$**Yx** (2)

, where M=diag($y_{k +}$), an M×M diagonal matrix containing the row
totals of **Y**. the dispersion of the species scores
δ=$\sum_{k = 1}^{M}{y_{k +}u^{2}/y_{+ +}}$ then can be calculate
through:
δ=$\mathbf{u}^{\mathbf{T}}$**Mu/**$\mathbf{x}^{\mathbf{T}}$**Nx**=$\mathbf{x}^{\mathbf{T}}\mathbf{Y}^{\mathbf{T}}\mathbf{M}^{\mathbf{- 1}}$**Yx/**$\mathbf{x}^{\mathbf{T}}$**Nx**

The problem of maximizing δ with respect to x has as solution the second
eigenvector of the eigenvalue equation

$\mathbf{Y}^{\mathbf{T}}\mathbf{M}^{\mathbf{- 1}}$**Yx** =λ**Nx** ,with
δ=λ (3)

This can be seen by noting that the first eigenvector is a trivial
solution (**x**=**1**; λ=1); because the second eigenvector is
orthogonal to the first eigenvector in the N metric, the second
eigenvector maximize δ subject to $\mathbf{1}^{\mathbf{T}}$**Nx=0.**
What is called the first eigenvector of CA in above reciprocal averaging
is thus the second eigenvector of (2), i.e. its first non-trivial
eigenvector. The second non-trivial eigenvector of (2) is similarly seen
to maximize δ, subject to being centered and to being orthogonal to the
first non-trivial eigenvector, and so on for subsequent axes. (2) can be
rewritten as

λ**x=**$\mathbf{N}^{\mathbf{- 1}}\mathbf{Y}^{\mathbf{T}}$**u** (4)

Equation (2)(4) form the ‘transition equations’ of CA. In words: the
species scores are weighted averages of the sites scores and the site
scores are proportional to weighted averages of the species scores.
