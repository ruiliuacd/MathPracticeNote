**2. The algebraic view of column vectors in the above ordination method is essentially a PCA projection**

Regard the matrix Y as composed of N M-dim factors (sites or individuals), the essential process of above least-square based ordination is to find a direction $\mathbf{δ}$, so that the projection of these factors onto $\mathbf{δ}$ as dispersed as possible.


$\mathbf{δ}_{1}^{T}$·($\mathbf{a}_{\mathbf{1}}$,...,$\mathbf{a}_{\mathbf{N}}$)=$\mathbf{x}_{\mathbf{1}}$=(x<sub>1</sub>,x<sub>2</sub>,...,x<sub>N</sub>), x<sub>i</sub> is the projection of each $\mathbf{a}_{\mathbf{i}}$ onto the PC1 axis $\mathbf{δ}_{1}$
