## Brief Outlining Coalescent Theory Principles (and Its Applications)

The coalescent theory is developed from the idea that when sample size n (individuals or gene copies no matter those copies are the same or bi-/multi- alleles) is much smaller than the population size N where the sample from. 

The possibility that all those gene copies came from different copies in the preceding generation is $G_{n,n}$(not coalescent in a generation). Under W-F model, the possibility that n samples came from i (i\<=n) individuals in last generation is: (ignoring mutation)

$G_{n,i}$=$\frac{S_{n}^{(i)}N_{\lbrack i\rbrack}}{N^{n}}$, where
$N_{\lbrack i\rbrack} = N(N - 1)\ldots(N - i + 1)$ , $S_{n}^{(i)}$ is
second Stirling number.  
  
Table 1. The probability $G_{n,i}$, that n=10 samples have i ancestors in the immediately previous generation for different values of N. Values < 10e-3 are omitted for readability.  

We can find that the possibility for i\<=n-2 has been too small and can be negligible when compared to i=n-1 or i=n when N is big.
i.e. $G_{n,n}$+$G_{n,n - 1}$ ≈ 1 and the possibility of i=n-1 is also
much smaller than n=i. Then we can make derivation for coalescent as illustration in [page 2](docs/ruiliu%20coalescent.pdf).  The waiting generations for one coalescent
event can be described as a geometry distribution: a coalescent event
occurred after t time interval

P{T=t}=${(1 - G_{n,n - 1})}^{(t - 1)}{G_{n,n - 1}}^{1} = \lambda\Delta t{(1 - \lambda\Delta t)}^{t/\Delta t}$,

assuming events occur at some rate λ per unit of time and $\delta$t
denote a tiny interval of continuous time. And it will eventually trends
to $e^{- \lambda t}$ according to Taylor series expansion,

$\lim_{\Delta t \rightarrow 0}(1 - \lambda\Delta t)^{t/\Delta t} = {{(e}^{- \lambda\Delta t})}^{t/\Delta\iota}$=$e^{- \lambda t}$.

Expected waiting time for one coalescent event of n samples:
E\[T<sub>n</sub>\]=1/$\lambda\ $(=1/$C_{n}^{2}$=2/n(n-1) when time unit
is N generations).

Similary, we can derive the possibility that the event do not happen during t interval for both coalescent, mutation, recombination events:  
P{no coalescence}= $e^{-λt}$； λ=$C_{n}^{2}$=$\sum_{i≠j}I_{ij}$;   lineages decrease by one    
P{no mutation}   = $e^{-θt}$，n=2; λ=θ=$4N_{e}μ$                 
P{no recombination}  =$e^{-kρt/2}$; ρ=$4N_{e}r$, λ=kρ/2        ;  lineages increase by one. Detailed in [page 3](docs/ruiliu%20coalescent.pdf)  

The coalescent theory has been widely used in demographic inference and
selection identifying. Such as \[4\] developed a balancing selection
identifying method utilizing sister species/populations based on
coalescent theory.
